```r
library(arrow)
library(fixest)
library(dplyr)
input_dir <- Sys.getenv("FDA_REGRESSION_DIR", "data/regression")
output_dir <- Sys.getenv("FDA_REGRESSION_OUTPUT_DIR", "outputs/regression")
dir.create(output_dir, recursive=TRUE, showWarnings=FALSE)
read_regression_input <- function(name) {
  path <- file.path(input_dir, name)
  if (!file.exists(path)) stop(paste("Missing regression input:", path, "See README.md for required files and fields."))
  result <- arrow::read_parquet(path)
  required <- c(independent_var, dependent_vars_ols, fixed_effects, control_vars)
  missing <- setdiff(required, names(result))
  if (length(missing)) stop(paste("Missing fields:", paste(missing, collapse=", ")))
  result
}
```

# Patents

```r
independent_var <- 'is_orange' 
dependent_vars_ols <- c('cit_pct', 'xi_real', 'num_claims') 
fixed_effects <- c('cpc_section','Year')
control_vars <- c('nau', 'nref')

df <- read_regression_input('reg1_patent.parquet')
df$Year <- as.factor(df$Year)
df$cpc_section <- as.factor(df$cpc_section)
```

```r
controls_formula <- paste(control_vars, collapse = ' + ')
fixed_effects_formula <- paste(fixed_effects, collapse = ' + ')

rhs_formula <- paste(independent_var, controls_formula, sep = ' + ')
full_rhs_formula <- paste(rhs_formula, fixed_effects_formula, sep = ' | ')

single_models_ols <- list()
for (dep_var in dependent_vars_ols) { 
    formula_str <- paste0(dep_var, ' ~ ', full_rhs_formula)
    model <- feols(as.formula(formula_str), data = df)
    model_name <- paste0("OLS_M_", dep_var) 
    single_models_ols[[model_name]] <- model
    cat(paste0("Model ", model_name, " (Dependent: ", dep_var, ") Complete.\n"))
}
etable(single_models_ols)
write.csv(etable(single_models_ols), file.path(output_dir, "reg_1_patents.csv"))
```
